```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1 пункт
k = 2
b = 4
X = np.linspace(0, 10, 67)
Y = k*X + b - 0.67*np.random.randn(*X.shape)
```

```python
# 2 пункт
k1, b1 = np.random.normal(size=(2))
Y_n = k1*X + b1
```

```python
# 2 пункт
plt.scatter(X, Y, color="green")
plt.plot(X, Y_n, color="red")
plt.show()
```

```python
K = np.array([k1,])
B = np.array([b1,])

# 2 пункт
def train_step(X, Y, border, lr=0.1):
    loss_old = (border[0]*X + border[1] - Y)**2
    loss_old = np.sum(loss_old)

    delta = np.random.normal(size=(2))
    new_border = border + lr*delta

    loss_new = (new_border[0]*X + new_border[1] - Y)**2
    loss_new = np.sum(loss_new)

    if loss_old > loss_new:
        border = new_border

    return border

# 5 пункт
for i in range(500):
    k1,b1 = train_step(X, Y, np.array((k1, b1)))
    K = np.concatenate((K, [k1]))
    B = np.concatenate((B, [b1]))
```

```python
# 5 пункт
plt.xlabel("k")
plt.ylabel("b")
plt.scatter(k1, b1, marker="x", label="Старт")
plt.scatter(k1, b1, marker="o", label="Финиш")
plt.legend()
plt.grid(True)
plt.plot(K, B)
plt.show()
```

```python
# 3 пункт
plt.scatter(X, Y, color="green")
plt.plot(X, k1*X + b1, color="red")
plt.show()
```
